# Type2: why A and T give the same circuit

For a homogeneous circular wire, real positive expansion point, and four elements $R_0,L_1,R_2,L_3$, this notebook verifies that the A and T field formulations give the same Type2 ladder. This is a continuum result; it does not establish equality on a finite mesh, arbitrary stage counts, or general three-dimensional geometries.

Let $x=r/a$, $z=\mu\sigma a^2s$ and
$$e=I_0(\sqrt z x)/I_0(\sqrt z),\qquad h=I_1(\sqrt z x)/(\sqrt z I_0(\sqrt z)).$$
Set $Y=2\pi h(1)$, $W=2\pi\int_0^1 xh^2dx=-Y'$. Unit-current fields $(e,h)/Y$ have total energy integral $1/Y$; dividing by that value recovers the unit-voltage fields $(e,h)$.

The common next stage is
$$E_2=e+c\partial_z e,\quad H_2=h+c\partial_z h,\quad c=Y/W=R_0/L_1.$$
It satisfies **both** $E_2(1)=1$ and $H_2(1)=0$. Thus the A and T boundary conditions select the same field, despite differently written solves. No boundary-condition change is required.

Uniqueness follows from $[x\delta E\delta H]_0^1=\int_0^1x((\delta E)^2+z(\delta H)^2)dx$ with regular axis and homogeneous boundary data.

The element formulas are
$$R_0=1/Y,\quad L_1=W/Y^2,\quad R_2=\left[2\pi\int_0^1x(E_2^2+zH_2^2)dx\right]^{-1},$$
$$H_3=R_2H_2+R_0h,\qquad L_3=2\pi\int_0^1xH_3^2dx.$$
The script independently extracts the same elements from the exact impedance Taylor series. The finite ladder matches through cubic order in $q=z-z_0$, not at every frequency. Resistance unit: $1/(\sigma a^2)$; inductance unit: $\mu$.

## Reproduce
Install a Wolfram runtime with `wolframscript` on PATH (or set `WOLFRAMSCRIPT`). The Python cell runs the `.wls` and requires all checks to pass. Stored output was generated with Wolfram 14.2. Numerical agreement below $10^{-28}$ is a finite-precision test, not an exact numerical proof. The 12 symbolic identities were checked exactly. The underlying derivation was independently reviewed with Claude Opus 5.5; subsequent publication packaging was checked by Codex.


```python
from pathlib import Path
import os, shutil, subprocess, tempfile, json
root = Path.cwd().parent if Path.cwd().name == "docs" else Path.cwd()
exe = os.environ.get("WOLFRAMSCRIPT") or shutil.which("wolframscript")
if not exe:
    raise RuntimeError("Install wolframscript or set WOLFRAMSCRIPT")
with tempfile.TemporaryDirectory() as tmp:
    output = Path(tmp) / "results.json"
    run = subprocess.run([exe, "-file", str(root / "mathematica/derive_type2_same_circuit.wls"), str(output)], capture_output=True, text=True)
    print("\n".join(line for line in run.stdout.splitlines() if not line.startswith("RESULT_FILE ")))
    if run.returncode:
        raise RuntimeError(run.stderr or "Wolfram validation failed")
    results = json.loads(output.read_text())
    assert all(results["symbolicChecks"].values())
    assert all(case["passed"] for case in results["cases"])
    print("All symbolic and numerical checks passed.")
```
